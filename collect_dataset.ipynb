```python
import cv2
import pandas as pd
import os
import mediapipe as mp

from mediapipe.tasks.python import vision
from mediapipe.tasks.python import BaseOptions

base_options = BaseOptions(
    model_asset_path="hand_landmarker.task"
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=1
)

landmarker = vision.HandLandmarker.create_from_options(options)

columns = []
for i in range(1, 22):
    columns.append(f"x{i}")
    columns.append(f"y{i}")
columns.append("label")

df = pd.DataFrame(columns=columns)

cap = cv2.VideoCapture(0)

print("===== DATASET COLLECTION STARTED =====")
print("m = MOVE | l = LEFT_CLICK | r = RIGHT_CLICK")
print("s = SCROLL | d = DRAG | x = STOP")
print("ESC = Exit")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.flip(frame, 1)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = landmarker.detect(mp_image)

    if result.hand_landmarks:
        hand_landmarks = result.hand_landmarks[0]

        landmarks = []
        for lm in hand_landmarks:
            landmarks.append(lm.x)
            landmarks.append(lm.y)

        key = cv2.waitKey(1) & 0xFF
        label = None

        if key == ord('m'):
            label = "MOVE"
        elif key == ord('l'):
            label = "LEFT_CLICK"
        elif key == ord('r'):
            label = "RIGHT_CLICK"
        elif key == ord('s'):
            label = "SCROLL"
        elif key == ord('d'):
            label = "DRAG"
        elif key == ord('x'):
            label = "STOP"

        if label and len(landmarks) == 42:
            landmarks.append(label)
            df.loc[len(df)] = landmarks
            print(f"Saved: {label}")

    cv2.putText(
        frame,
        "Press key to save gesture | ESC to exit",
        (10, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 0),
        2
    )

    cv2.imshow("Dataset Collection", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

os.makedirs("dataset", exist_ok=True)
df.to_csv("dataset/gesture_data.csv", index=False)

cap.release()
cv2.destroyAllWindows()
landmarker.close()

print("===== DATASET SAVED SUCCESSFULLY =====")
print(f"Total samples: {len(df)}")

```
